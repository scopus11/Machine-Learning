```python
import pandas as pd
from sklearn.metrics import confusion_matrix, accuracy_score, precision_score, recall_score
```

```python
df= pd.read_excel(path + '/dataset/prediksi_kelulusan_dataset.xlsx')
```

```python
df.head()
```

```python
df.info()
```

```python
# Ubah label ke biner
# Lulus = 1, Tidak Lulus = 0
df['Actual_bin'] = df['Hasil Sebenarnya'].map({"Lulus":1, "Tidak Lulus":0})
df['Predicted_bin'] = df['Hasil Prediksi'].map({"Lulus":1, "Tidak Lulus":0})
y_true = df['Actual_bin']
y_pred = df['Predicted_bin']
```

```python
# Confusion Matrix
cm = confusion_matrix(y_true, y_pred)
print("=== Confusion Matrix ===")
print(cm)
```

```python
# Hitung metrik
accuracy = accuracy_score(y_true, y_pred)
precision = precision_score(y_true, y_pred)
recall = recall_score(y_true, y_pred)
```

```python
print("\n=== Evaluasi Model ===")
print(f"Accuracy : {accuracy:.2f}")
print(f"Precision: {precision:.2f}")
print(f"Recall   : {recall:.2f}")
```

```python
# Penjelasan tambahan buat membantu memahami output.
print("\n== Interpretasi ==")
print("- Accuracy mengukur prediksi benar dari seluruh data.")
print("- Precision menunjukkan kualitas prediksi 'Lulus'.")
print("- Recall menunjukkan kemampuan model menemukan semua mahasiswa yang benar-benar 'Lulus'.")
```
